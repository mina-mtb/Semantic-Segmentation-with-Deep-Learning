#   GPU Acceleration Verification

This notebook verifies that your **NVIDIA GTX 1650** is correctly detected by TensorFlow using the **DirectML** plugin.

```python
import tensorflow as tf
import time

print(f"TensorFlow version: {tf.__version__}")

# Check for physical devices detected by DirectML
gpus = tf.config.list_physical_devices('GPU')
if gpus:
    print("\n  SUCCESS! GPU(s) detected:")
    for gpu in gpus:
        print(f"    {gpu}")
else:
    print("\n  ERROR: No GPU detected. Make sure you restarted the kernel.")
```

##   Speed Test
Let's compare a large matrix multiplication on CPU vs GPU.

```python
size = 5000

print(f"Benchmarking matrix multiplication ({size}x{size})...")

# --- CPU Test ---
with tf.device('/CPU:0'):
    a = tf.random.normal([size, size])
    b = tf.random.normal([size, size])
    start = time.time()
    c = tf.matmul(a, b)
    cpu_time = time.time() - start
    print(f"CPU Time: {cpu_time:.4f} seconds")

# --- GPU Test ---
try:
    with tf.device('/GPU:0'):
        a = tf.random.normal([size, size])
        b = tf.random.normal([size, size])
        # Warm up
        _ = tf.matmul(a[0:10, 0:10], b[0:10, 0:10])
        
        start = time.time()
        c = tf.matmul(a, b)
        gpu_time = time.time() - start
        print(f"GPU Time: {gpu_time:.4f} seconds")
        
        speedup = cpu_time / gpu_time
        print(f"\n  GPU is {speedup:.1f}x faster than CPU!")
except Exception as e:
    print(f"\n  GPU Test failed: {e}")
```
